```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['figure.figsize'] = [7.00, 3.50]
plt.rcParams['figure.autolayout'] = True
```

### Weekly_Sales distribution.

```python
fig, axes = plt.subplots(1, 2, figsize=(14.00, 3.50))
sns.histplot(data=clean_d, x='Weekly_Sales', bins=50, kde=True, ax=axes[0])
sns.boxplot(data=clean_d, y='Weekly_Sales', ax=axes[1])
plt.show()
```

The data is right-skewed.

### Linear variable vs Weekly_Sales

```python
fig, axes = plt.subplots(2, 2, figsize=(14.00, 7.00))
sns.scatterplot(data = clean_d, 
                x = 'Temperature', 
                y = 'Weekly_Sales', 
                hue = 'Holiday_Flag', 
                alpha = 0.6, 
                ax = axes[0, 0])
sns.scatterplot(data = clean_d, 
                x = 'Fuel_Price', 
                y = 'Weekly_Sales', 
                hue = 'Holiday_Flag', 
                alpha = 0.6, 
                ax = axes[0, 1])
sns.scatterplot(data=clean_d, 
                x = 'CPI', 
                y = 'Weekly_Sales', 
                hue = 'Holiday_Flag', 
                alpha = 0.6, 
                ax = axes[1, 0])
sns.scatterplot(data = clean_d, 
                x = 'Unemployment', 
                y = 'Weekly_Sales', 
                hue = 'Holiday_Flag', 
                alpha = 0.6, 
                ax = axes[1, 1])
plt.show()
```

1. Temperature:  
0-20 and 100 ,sales are low.20-90, sales are plenty and they're evently distributed.  
2. Fuel_Price:  
Range from 2.50 to 4.25, sales are plenty and they're evently distrubuted. 4.50, sales are merely zero.  
3. CPI:  
<140 and 180-220, sales are plenty. No sales in the remaining ranges.
4. Unemployment:  
Most sales are concentrated where unemployment is between 6 and 9.
