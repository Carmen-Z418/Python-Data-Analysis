## Introduction

These data are expressed in constant international dollars to adjust for inflation and differences in the cost of living across countries.  
Depending on the country and year, the data refer to income after taxes and transfers, or to per capita consumption.   

Many people, both today and in the past, have no monetary income. These data account for this by including estimated values of non-market income, such as food grown for their own use by subsistence farmers.
