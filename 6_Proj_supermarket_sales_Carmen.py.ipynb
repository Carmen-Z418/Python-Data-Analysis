# Project: Supermarket Sales

## Introduction

This dataset contains supermarket sales data.  
The description of variables:
- Store ID：The(index)ID of a specific store
- Store_Area：The physical area of the store
- Items_Available：The number of items avaliable in the store
- Daily_Customer_Count：The average number of customers visiting the store per manth
- Store_Sales：The sales revenue of the store(in USD)

## Analyze objective

1. Analyze the relationship between store size and sales revenue.  
2. Analyze the relationship between daily customer count and sales revenue.  
3. Analyze the relationship between store size and daily customer.

## Explore data

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_palette('pastel')
```

### Store_area vs Store_sales

```python
fig, axes = plt.subplots(1, 2)
sns.histplot(data=ori_d, x='Store_Area', ax=axes[0])
sns.boxplot(data=ori_d, y='Store_Area', ax=axes[1])
plt.show()
```

`Store_Area` follows a normal distribution, with most data concentrated between 1,000 to 2,000.

```python
sns.scatterplot(data=ori_d, x='Store_Area', y='Store_Sales')
plt.show()
```

There is no clear linear relationship between store area and sales revenue.

### Daily_customer_counts vs Store_sales

```python
sns.scatterplot(data=ori_d, x='Daily_Customer_Count', y='Store_Sales')
plt.show()
```

Daily customer count doesn't show a strong correlation with sales revenue.

### Store_area vs Daily_customer_counts

```python
sns.scatterplot(data=ori_d, x='Daily_Customer_Count', y='Store_Area', ci=None)
plt.show()
```

Daily customer count doesn't show a strong correlation with store area.
